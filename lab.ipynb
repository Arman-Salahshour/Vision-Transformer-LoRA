### Download the data

```python
import os
import urllib.request
import tarfile

# If the "data" directory does not exist, create it and download the ImageWoof dataset
if not os.path.exists("data"):
    data_dir = os.path.join(os.getcwd(), "data")
    os.makedirs(data_dir, exist_ok=True)
    url = "https://s3.amazonaws.com/fast-ai-imageclas/imagewoof2.tgz"
    file_path = os.path.join(data_dir, "imagewoof2.tgz")

    # Download the dataset
    print("Downloading ImageWoof dataset ")
    urllib.request.urlretrieve(url, file_path)

    # Extract the dataset
    print("Extracting dataset...")
    with tarfile.open(file_path, "r:gz") as tar:
        tar.extractall(path=data_dir)
```
